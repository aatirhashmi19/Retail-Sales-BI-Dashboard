Objective:

Store the cleaned retail sales dataset inside PostgreSQL and perform SQL analysis.

```python
import pandas as pd
from sqlalchemy import create_engine
```

```python
df = pd.read_csv("data/processed/superstore_cleaned.csv")
```

```python
DATABASE_URL = "postgresql://postgres:22332233@localhost:5432/retail_sales_db"

engine = create_engine(DATABASE_URL)

print("Connected Successfully")
```

```python
df.to_sql(
    "superstore",
    engine,
    if_exists="replace",
    index=False
)

print("Dataset Uploaded Successfully")
```

```python
query = """
SELECT table_name
FROM information_schema.tables
WHERE table_schema = 'public';
"""

pd.read_sql(query, engine)
```

```python
query = """
SELECT *
FROM superstore
LIMIT 5;
"""

pd.read_sql(query, engine)
```


# Total Sales




```python
query = """
SELECT
SUM("Sales") AS Total_Sales
FROM superstore;
"""

pd.read_sql(query, engine)
```

# total Profit

```python
query = """
SELECT
SUM("Profit") AS Total_Profit
FROM superstore;
"""

pd.read_sql(query, engine)
```

# Sales by Category

```python
query = """
SELECT
"Category",
SUM("Sales") AS Total_Sales

FROM superstore

GROUP BY "Category"

ORDER BY Total_Sales DESC;
"""

pd.read_sql(query, engine)
```

# Profit by Category

```python
query = """
SELECT
"Category",
SUM("Profit") AS Total_Profit

FROM superstore

GROUP BY "Category"

ORDER BY Total_Profit DESC;
"""

pd.read_sql(query, engine)
```

# Sales by Region

```python
query = """
SELECT
"Region",
SUM("Sales") AS Total_Sales

FROM superstore

GROUP BY "Region"

ORDER BY Total_Sales DESC;
"""

pd.read_sql(query, engine)
```

# Top 10 Customers

```python
query = """
SELECT
"Customer Name",
SUM("Sales") AS Total_Sales

FROM superstore

GROUP BY "Customer Name"

ORDER BY Total_Sales DESC

LIMIT 10;
"""

pd.read_sql(query, engine)
```

# Top 10 Products

```python
query = """
SELECT
"Product Name",
SUM("Sales") AS Total_Sales

FROM superstore

GROUP BY "Product Name"

ORDER BY Total_Sales DESC

LIMIT 10;
"""

pd.read_sql(query, engine)
```

# Monthly Sale

```python
query = """
SELECT
"Year",
"Month",
SUM("Sales") AS Monthly_Sales

FROM superstore

GROUP BY "Year","Month"

ORDER BY "Year","Month";
"""

pd.read_sql(query, engine)
```

# Average Shipping Days

```python
query = """
SELECT
AVG("Shipping Days") AS Average_Shipping_Days

FROM superstore;
"""

pd.read_sql(query, engine)
```

# Highest Profit State

```python
query = """
SELECT
"State",
SUM("Profit") AS Profit

FROM superstore

GROUP BY "State"

ORDER BY Profit DESC

LIMIT 10;
"""

pd.read_sql(query, engine)
```

# Highest Discount Products

```python
query = """
SELECT
"Product Name",
AVG("Discount") AS Average_Discount

FROM superstore

GROUP BY "Product Name"

ORDER BY Average_Discount DESC

LIMIT 10;
"""

pd.read_sql(query, engine)
```
