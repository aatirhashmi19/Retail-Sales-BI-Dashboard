## Objective

Prepare aggregated datasets for creating an interactive Power BI dashboard.

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("data/processed/superstore_cleaned.csv")

df["Order Date"] = pd.to_datetime(df["Order Date"])

df.head()
```

# KPI Table

```python
kpi = pd.DataFrame({

    "Metric":[
        "Total Sales",
        "Total Profit",
        "Total Orders",
        "Total Customers",
        "Average Order Value"
    ],

    "Value":[
        df["Sales"].sum(),
        df["Profit"].sum(),
        df["Order ID"].nunique(),
        df["Customer ID"].nunique(),
        df["Sales"].sum()/df["Order ID"].nunique()
    ]

})

kpi
```

# Sales by Category

```python
category_summary = (
    df.groupby("Category")
    .agg(
        Sales=("Sales","sum"),
        Profit=("Profit","sum"),
        Orders=("Order ID","nunique")
    )
    .reset_index()
)

category_summary
```

# Sales by Region

```python
region_summary = (
    df.groupby("Region")
    .agg(
        Sales=("Sales","sum"),
        Profit=("Profit","sum")
    )
    .reset_index()
)

region_summary
```

# Customer Summary

```python
customer_summary = (
    df.groupby("Customer Name")
    .agg(
        Sales=("Sales","sum"),
        Profit=("Profit","sum"),
        Orders=("Order ID","nunique")
    )
    .reset_index()
)

customer_summary.head()
```

# Product Summary

```python
product_summary = (
    df.groupby("Product Name")
    .agg(
        Sales=("Sales","sum"),
        Profit=("Profit","sum"),
        Quantity=("Quantity","sum")
    )
    .reset_index()
)

product_summary.head()
```

# Monthly Sales Summary

```python
monthly_summary = (
    df.groupby(["Year","Month","Month Name"])
    .agg(
        Sales=("Sales","sum"),
        Profit=("Profit","sum")
    )
    .reset_index()
)

monthly_summary.head()
```

# State Summary

```python
state_summary = (
    df.groupby("State")
    .agg(
        Sales=("Sales","sum"),
        Profit=("Profit","sum")
    )
    .reset_index()
)

state_summary.head()
```

# Export Dashboard Files

```python
kpi.to_csv(
    "dashboard/KPI.csv",
    index=False
)

category_summary.to_csv(
    "dashboard/Category_Summary.csv",
    index=False
)

region_summary.to_csv(
    "dashboard/Region_Summary.csv",
    index=False
)

customer_summary.to_csv(
    "dashboard/Customer_Summary.csv",
    index=False
)

product_summary.to_csv(
    "dashboard/Product_Summary.csv",
    index=False
)

monthly_summary.to_csv(
    "dashboard/Monthly_Summary.csv",
    index=False
)

state_summary.to_csv(
    "dashboard/State_Summary.csv",
    index=False
)

print("Dashboard files exported successfully!")
```
